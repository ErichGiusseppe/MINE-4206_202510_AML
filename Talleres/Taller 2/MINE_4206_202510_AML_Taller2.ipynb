## Explicación de cada métrica  

- **Precisión (Precision):** Indica la proporción de instancias clasificadas como positivas que realmente son positivas.  
  - Se calcula como:  
    Precisión = $ \frac{TP}{TP + FP} $
  - Donde:  
    - \( TP \) (True Positives) son los verdaderos positivos.  
    - \( FP \) (False Positives) son los falsos positivos.  

- **Recall (Sensibilidad):** Mide la capacidad del modelo para encontrar todas las instancias positivas.  
  - Se calcula como:  
    Recall = $ \frac{TP}{TP + FN} $
  - Donde:  
    - \( TP \) son los verdaderos positivos.  
    - \( FN \) (False Negatives) son los falsos negativos.  

- **F1-Score:** Es la media armónica entre precisión y recall, proporcionando un balance entre ambas métricas.  
  - Se calcula como:  
    F1 = $2 \times \frac{\text{Precisión} \times \text{Recall}}{\text{Precisión} + \text{Recall}}$

Todas las metricas antes presentadas serán calculadas en la version *weighted avg*, dado que hay un desbalance de clases para la clase a predecir. por otro lado para suplir el problema de balance se va dar más peso a la clase que tiene menos registros, esto para asegurar que el modelo no solo se centre en predecir bien la clase qeu tiene una mayor cantidad de registros.

Nota: Usualmente para se utiliza el F1 pero si el problema requiere un foco específico se puede usar recall o precisión.

```python
def plot_confusion_matrix(
    y_true, y_pred, classes, normalize=False, title=None, cmap=plt.cm.Blues, size=(4, 4)
):
    """
    This function prints and plots the confusion matrix.
    Normalization can be applied by setting `normalize=True`.
    """
    if not title:
        if normalize:
            title = "Normalized confusion matrix"
        else:
            title = "Confusion matrix, without normalization"
    cm = confusion_matrix(y_true, y_pred, labels=["Stable","At Risk"]) #  

    if normalize:
        cm = cm.astype("float") / cm.sum(axis=1)[:, np.newaxis]
        print("Normalized confusion matrix")
    else:
        print("Confusion matrix, without normalization")

    fig, ax = plt.subplots(figsize=size)
    im = ax.imshow(cm, interpolation="nearest", cmap=cmap)
    ax.figure.colorbar(im, ax=ax)
    # We want to show all ticks...
    ax.set(
        xticks=np.arange(cm.shape[1]),
        yticks=np.arange(cm.shape[0]),
        # ... and label them with the respective list entries
        xticklabels=classes,
        yticklabels=classes,
        title=title,
        ylabel="True label",
        xlabel="Predicted label",
    )

    plt.setp(ax.get_xticklabels(), rotation=45, ha="right", rotation_mode="anchor")

    fmt = ".2f" if normalize else "d"
    thresh = cm.max() / 2.0
    for i in range(cm.shape[0]):
        for j in range(cm.shape[1]):
            ax.text(
                j,
                i,
                format(cm[i, j], fmt),
                ha="center",
                va="center",
                color="white" if cm[i, j] > thresh else "black",
            )
    fig.tight_layout()
    return ax
```

```python
def display_multiple_models_roc_curve(models, X, y):
    for m in models:
        model = m["model"]  
        y_pred = model.predict(X)  
        fpr, tpr, thresholds = roc_curve(y, model.predict_proba(X)[:, 1])
        auc = roc_auc_score(y, y_pred)
        plt.plot(fpr, tpr, label="%s ROC (area = %0.2f)" % (m["label"], auc))
    plt.plot([0, 1], [0, 1], "r--")
    plt.xlim([0.0, 1.0])
    plt.ylim([0.0, 1.05])
    plt.xlabel("1-Specificity(False Positive Rate)")
    plt.ylabel("Sensitivity(True Positive Rate)")
    plt.title("Receiver Operating Characteristic")
    plt.legend(loc="lower right")
    plt.show()  
```

```python

class CustomPreprocessor(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None):
        return self

    def transform(self, X):
        X = X.copy()
        # X.drop(columns=["Month"])
        X["Average Fish Weight (g)"] = X["Average Fish Weight (g)"].round(2)
        X["Survival Rate (%)"] = X["Survival Rate (%)"].round(2)
        # X = X[X["Disease Occurrence"] % 1 == 0]
        # X["Temperature (°C)"] = X["Temperature (°C)"].round(2)
        X["Dissolved Oxygen (mg/L)"] = X["Dissolved Oxygen (mg/L)"].round(2)
        X["pH"] = X["pH"].round(2)
        X["Turbidity (NTU)"] = X["Turbidity (NTU)"].round(2)
        X["Oxygenation Interventions"] = X["Oxygenation Interventions"].astype(int)
        return X


numericos = [
    col
    for col in train.drop(
        columns=["Thermal Risk Index", "Health Status", "Temperature (°C)"]
    ).select_dtypes(include=["float64", "int64", "int32"])
    if "latitud" not in col and "longitud" not in col
]
categoricas = [
    col
    for col in train.drop(
        columns=["Thermal Risk Index", "Health Status", "Temperature (°C)"]
    ).select_dtypes(include=["object", "category"])
]
cat_transformer = Pipeline([("encoder", OneHotEncoder(handle_unknown="ignore"))])
num_transformer2 = Pipeline([("scaler", StandardScaler())])

preprocessor = Pipeline(
    [
        ("custom_transform", CustomPreprocessor()),
        (
            "column_transform",
            ColumnTransformer(
                [
                    ("num2", num_transformer2, numericos),
                    ("cat", cat_transformer, categoricas),
                ]
            ),
        ),
    ]
)
```

## SVC

```python
estimators_bm = [
    ("transform", preprocessor),
    ("classifier", SVC(kernel="rbf", probability=True, random_state=77)),
]
pipe_bm = Pipeline(estimators_bm)

param_space = {
    "transform__column_transform__num2__scaler": Categorical(
        [MinMaxScaler(), StandardScaler(), RobustScaler(), "passthrough"]
    ),
    "classifier__C": (1e-6, 100.0, "log-uniform"),
    "classifier__gamma": Categorical(["scale", "auto"]),
}

bayes_search_svc = BayesSearchCV(
    pipe_bm,
    param_space,
    n_iter=50,
    cv=3,
    scoring="f1_weighted",
    n_jobs=-1,
    verbose=2,
    random_state=77,
)

bayes_search_svc.fit(X_train, Y_train)

print("Mejores hiperparámetros encontrados:")
print(bayes_search_svc.best_params_)
```

### **Análisis de resultados:**

### ¿Qué puedes decir de los valores de las métricas recall y precisión para cada una de las clases en cada modelo? ¿Cuál de estás métricas consideras que es más importante con base en la descripción del problema?

En general, ambas métricas son similares, pero evalúan aspectos distintos del rendimiento del modelo.

Precisión responde a la pregunta: "De todas las predicciones positivas, ¿cuántas realmente lo son?" En este caso, nos indica qué proporción de los casos clasificados como "en riesgo" realmente están en riesgo.

Recall responde a la pregunta: "De todas las instancias realmente positivas, ¿cuántas fueron detectadas?" Es decir, de todos los casos que realmente están en riesgo, ¿cuántos fueron correctamente identificados por el modelo?

Dado que el escenario más crítico es la identificación y abordaje de problemas de salud en los peces, es más peligroso que el modelo pase por alto un caso de riesgo (falso negativo) a que clasifique erróneamente un caso sano como "en riesgo" (falso positivo). En otras palabras, es preferible que el modelo emita una alerta innecesaria a que deje pasar un caso que realmente representa un problema de salud.

En cuanto a los resultados, el mejor modelo es el de árboles de decisión, ya que presenta métricas superiores en todas las clases y clasifica los casos de manera perfecta. Por lo tanto, nos quedamos con este modelo. 

### ¿Consideras que el rendimiento de los modelos es adecuado? Si no es así, ¿cómo podrían mejorarse los resultados?

El modelo de SVM es adecuado, pero su desempeño podría mejorarse ajustando el umbral de clasificación (threshold), siempre que exista un punto donde ambas clases se clasifiquen de manera más equilibrada. En modelos basados en probabilidades, modificar este umbral puede optimizar los resultados generales, especialmente si se busca minimizar los falsos negativos.

Desde una perspectiva de negocio, sería preferible reducir el umbral de decisión, ya que el costo de clasificar erróneamente un caso positivo como negativo (falso negativo) suele ser mayor que el de un falso positivo. Ajustar este umbral podría ayudar a mejorar la clasificación en escenarios críticos.

Es importante tener en cuenta que, por defecto, SVM **NO** estima probabilidades. Para habilitar esta funcionalidad, se debe activar la opción probability=True al entrenar el modelo. Luego, se puede determinar el umbral óptimo analizando la relación entre precisión y recall, utilizando la curva ROC como referencia.
